```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import time

img = cv2.imread("C:/Users/vinay/Downloads/Capture.jpg")
# Downscale image 50% → 4x faster processing
img = cv2.resize(img, (img.shape[1] // 2, img.shape[0] // 2))
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cimg = img.copy()

start = time.time()
# Optimized HoughCircles: dp=1.5 (vs 3), higher param2 (filter weak circles), minRadius=10
circles = cv2.HoughCircles(image=gray, method=cv2.HOUGH_GRADIENT, dp=1.5,
                            minDist=80, param1=100, param2=50, maxRadius=150, minRadius=10)
elapsed_ms = (time.time() - start) * 1000
print(f"HoughCircles time: {elapsed_ms:.4f} ms")
```

```python
if circles is not None:
    circles = np.uint16(np.around(circles))
    # Vectorized drawing: use broadcasting instead of loop
    for circle in circles[0]:
        cv2.circle(cimg, (circle[0], circle[1]), circle[2], (0, 255, 0), 2)
        cv2.circle(cimg, (circle[0], circle[1]), 2, (0, 0, 255), 3)
    num_circles = len(circles[0])
    print(f"Number of circles detected: {num_circles}")
else:
    print("No circles detected")

plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(cimg, cv2.COLOR_BGR2RGB))
plt.title(f"Circles Detected in {elapsed_ms:.4f} ms")
plt.axis('off')
plt.show()
```
